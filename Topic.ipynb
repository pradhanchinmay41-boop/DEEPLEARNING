###  Activation function 
* Activation function will implement in Hidden layer and output layer .
* Main purpose of Activation function is to introduce non-linearity .
* Sigmoid activation function(apply on output layer and it adjust outlier (Range=0 to 1))
* tanh activation(apply on o/p layer (Range = -1 to 1))
* Relu(Rectifier function) activation function (Formula (max(x,o))) here you will never get any -ve value. it apply on only hidden layer(RAnge = 0 to infinity).
* Leeky relu (While the data has any -ve values then we apply leeky relu). it alsoapply in hidden liyer .(Range = -1 to infinity)
* softmax(is  apply on o/p layer (* if thereis binary class thn apply sigmoidand if there is multiclass thn apply Softmax)).. the main advantage of using this is the output is in probabilities range (0 to 1).

* N.B : Except  softmax all  function are apply on binnary classification .
* we not apply every activation one time .

* Inteview Questions : Sigmoid fun vs tanh fun   and  relu vs leeky relu

### Formula : 
 * new weight = old weight - learningrate * (derivation of loss/derivation of old weight)

* Simple example

If:

    old weight = 0.5
    learning rate = 0.1
    gradient = 0.2

Then:


    new weight = 0.5 - (0.1 × 0.2)

# ANN MODEL BUILDING ARCHITECTURE

```python
ann = tf.keras.models.Sequential() #establishing layer one after another
```

* Here we  built a neutral network 
    * tf = backend 
    * keras = frontend
    * sequential means establish of the layers(i.e. created layer in memory)


```python
ann.add(tf.keras.layers.Dense(units=10)) # input layer
```

* unit= 6 (i.e. 6 neurons in i/p layers)
* Dense is connected the neurons to the layers .


```python
ann.add(tf.keras.layers.Dense(units=10, activation='relu')) # Hidden layer1
```

```python
ann.add(tf.keras.layers.Dense(units=5, activation='relu')) #2nd hidden layer
```

```python
ann.add(tf.keras.layers.Dense(units=1, activation='sigmoid')) #output layer
```

#### optimizer in Ann:
 * socastic gradient descend
 * adam
 * adagrad
 * adadelta
 * adamax
 * rmsprop

#### Loss :
  * Loss (or Loss Function) measures how far the model's prediction is from the actual answer.

  * Low loss → Predictions are close to actual values.
  * High loss → Predictions are far from actual values.
  * The goal of training is to minimize the loss.
  * regression loss --- mae,mse,logloss
  * classification  loss --- 
                            
                    * binary cross entropy(Use when there are 2 classes only.)

                   * category cross entropy(Use when there are more than 2 classes and labels are one-hot encoded.)

                   * sparse category cross entropy(Use when there are more than 2 classes and labels are integers, not one-hot encoded.)

```python
ann.compile(optimizer = 'adam', loss = 'binary_crossentropy', metrics = ['accuracy'])
```

```python

ann.fit(X_train,y_train,batch_size=32,epochs=50)
```

```python
y_pred = ann.predict(X_test)
y_pred = (y_pred > 0.5)
print(np.concatenate((y_pred.reshape(len(y_pred),1), y_test.reshape(len(y_test),1)),1))
```

```python
from sklearn.metrics import confusion_matrix
cm = confusion_matrix(y_test, y_pred)
print(cm)
```

```python
from sklearn.metrics import accuracy_score
ac = accuracy_score(y_test, y_pred)
print(ac)
```

# RNN
* It is advance NLP. it work on text data .
* In ANN - input layer -- Hidden layer -- output layer
* In CNN - input matrix -- filter -- output matrix -- maxpool - flatten -- fully connected .
* In RNN  -  input -- text preprocessing -- output 
* RNN also called sequence to sequence model coz everytime  it store previous output(not lose anything ) .
* AT last we should use softmax coz it always gone to be multiclass classification .
* RNN mainly work in 2 concept 
                     * 1. LSTM(long short term memory)
                     * 2. GRU(gated recurrant unit)

* LSTM - lstm have 3 gates 
                    * 1. input gate
                      * memory cell
                    * 3. output gate
                    * 4. Forget gate
* Live example : Gmail

* GRU have 2 gates 
             * 1. update gate(input + memory + output)
             * 2. reset gate (reset to new gmail id or....) 

* Gru is more advance but 99% lstm only use .

* NLP pre-trained model are BERT or GPT .
* computer vision  pretrained model are restnet /imagenet .
